```python
# ===================================
# VERIFICADOR DE TEMPERATURA
# ===================================

# Define o limite de temperatura (em graus Celsius)
limite = 30

# Solicita ao usuário que digite a temperatura atual
# float() permite usar números com casas decimais
temperatura = float(input("Digite a temperatura atual (°C): "))

# Verifica se a temperatura excede o limite definido
if temperatura > limite:
    # Se for maior, exibe um alerta
    print("⚠️ Alerta! Temperatura muito alta!")
    print("Temperatura registrada:", temperatura, "°C")
else:
    # Caso contrário, informa que está normal
    print("✅ Temperatura dentro do limite seguro.")
    print("Temperatura registrada:", temperatura, "°C")
```
