```python
# ==========================================
# SISTEMA DE MONITORAMENTO SIMPLES
# ==========================================

import random  # Usado para simular valores aleatórios
import time    # Usado para controlar o tempo entre as leituras

# Define o valor limite crítico (ex: 80)
LIMITE_CRITICO = 80

# ==========================================
# FUNÇÃO PARA GERAR MÉTRICA SIMULADA
# ==========================================
def gerar_metrica():
    """
    Simula a leitura de uma métrica (ex: CPU, temperatura, uso de memória).
    Retorna um valor aleatório entre 0 e 100.
    """
    return random.randint(0, 100)

# ==========================================
# FUNÇÃO PARA VERIFICAR A CONDIÇÃO
# ==========================================
def verificar_metrica(valor):
    """
    Verifica se o valor da métrica ultrapassa o limite crítico.
    Retorna True se for crítico, False caso contrário.
    """
    return valor > LIMITE_CRITICO

# ==========================================
# FUNÇÃO PARA ENVIAR ALERTA
# ==========================================
def enviar_alerta(valor):
    """
    Exibe um alerta no console informando que o valor está acima do limite.
    """
    print("⚠️ ALERTA CRÍTICO!")
    print(f"Valor da métrica muito alto: {valor}")
    print("Ação recomendada: verificar o sistema!\n")

# ==========================================
# LOOP PRINCIPAL
# ==========================================
def monitorar():
    """
    Executa o monitoramento continuamente.
    A cada ciclo:
    - Gera uma métrica
    - Verifica se é crítica
    - Dispara alerta se necessário
    """
    print("🚀 Iniciando monitoramento...\n")

    while True:
        # Gera uma nova métrica
        valor = gerar_metrica()

        # Mostra o valor atual
        print(f"Valor atual da métrica: {valor}")

        # Verifica se ultrapassou o limite
        if verificar_metrica(valor):
            enviar_alerta(valor)

        # Aguarda 2 segundos antes da próxima leitura
        time.sleep(2)

# ==========================================
# INÍCIO DO PROGRAMA
# ==========================================
if __name__ == "__main__":
    monitorar()
```
