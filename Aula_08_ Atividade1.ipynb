```python
from pathlib import Path

def organizar_pasta_por_categorias(pasta_origem, ignorar_extensoes=None):
    """
    Organiza arquivos de uma pasta agrupando-os por categorias lógicas
    (Imagens, Documentos, etc.) usando a biblioteca pathlib.
    """
    if ignorar_extensoes is None:
        ignorar_extensoes = set()
    else:
        # Garante que as extensões estejam em minúsculo e sem pontos extras
        ignorar_extensoes = {ext.lower().strip() for ext in ignorar_extensoes}

    caminho_base = Path(pasta_origem)

    if not caminho_base.exists() or not caminho_base.is_dir():
        print(f"✖ Diretorio inválido ou não encontrado: {pasta_origem}")
        return

    # Mapeamento de categorias (você pode customizar como preferir)
    CATEGORIAS = {
        "Imagens": [".jpg", ".jpeg", ".png", ".gif", ".bmp", ".svg", ".webp"],
        "Documentos": [".pdf", ".docx", ".doc", ".txt", ".xlsx", ".pptx", ".csv"],
        "Videos": [".mp4", ".mkv", ".avi", ".mov", ".flv"],
        "Compactados": [".zip", ".rar", ".7z", ".tar", ".gz"],
        "Audio": [".mp3", ".wav", ".flac", ".ogg"]
    }

    # Inverte o dicionário para facilitar a busca rápida da extensão -> categoria
    mapa_extensoes = {}
    for categoria, extensoes in CATEGORIAS.items():
        for ext in extensoes:
            mapa_extensoes[ext] = categoria

    resumo = {}
    print(f"\n🚀 Iniciando a triagem em: {caminho_base.resolve()}\n")

    # Itera apenas sobre os arquivos diretamente na pasta (ignora subpastas)
    for arquivo in caminho_base.iterdir():
        if arquivo.is_dir():
            continue

        ext = arquivo.suffix.lower()

        # Verifica se deve ignorar
        if ext in ignorar_extensoes:
            print(f"⚠ Ignorado por filtro: {arquivo.name}")
            continue

        # Determina a pasta de destino com base na categoria ou na própria extensão
        if ext == "":
            nome_pasta_destino = "Sem_Extensao"
        else:
            nome_pasta_destino = mapa_extensoes.get(ext, f"Outros_{ext[1:].upper()}")

        pasta_destino = caminho_base / nome_pasta_destino
        pasta_destino.mkdir(exist_ok=True)

        novo_caminho = pasta_destino / arquivo.name

        # Evita colisões de nomes adicionando um sufixo incremental
        contador = 1
        while novo_caminho.exists():
            novo_caminho = pasta_destino / f"{arquivo.stem}_{contador}{ext}"
            contador += 1

        try:
            # Move o arquivo de forma limpa usando pathlib
            arquivo.rename(novo_caminho)

            # Contabiliza no resumo
            resumo[nome_pasta_destino] = resumo.get(nome_pasta_destino, 0) + 1
            print(f"✔ {arquivo.name} ➔  📁 {nome_pasta_destino}")

        except Exception as e:
            print(f"✖ Falha ao mover {arquivo.name}: {e}")

    # Exibição do relatório final
    print("\n📊 Resumo da Organização:")
    if not resumo:
        print(" - Nenhum arquivo foi movido.")
    for pasta, qtd in resumo.items():
        print(f" 📁 {pasta}: {qtd} arquivo(s)")

    print("\n✨ Tudo pronto!")


# --- Exemplo de uso ---
if __name__ == "__main__":
    # Exemplo ignorando arquivos .tmp e .log
    organizar_pasta_por_categorias("./arquivos_para_organizar", ignorar_extensoes=[".tmp", ".log"])
```
