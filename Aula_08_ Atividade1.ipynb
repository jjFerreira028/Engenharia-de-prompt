```python
# ==========================================
# ORGANIZADOR DE ARQUIVOS POR EXTENSÃO
# ==========================================

import os   # Biblioteca para manipular arquivos e pastas
import shutil  # Biblioteca para mover arquivos

# ===============================
# DEFINA OS CAMINHOS DAS PASTAS
# ===============================

# Pasta onde estão os arquivos desorganizados
pasta_origem = "sua/pasta/origem"

# Pasta onde os arquivos serão organizados
pasta_destino = "sua/pasta/destino"

# ===============================
# FUNÇÃO PRINCIPAL
# ===============================

# Lista todos os arquivos da pasta de origem
for arquivo in os.listdir(pasta_origem):

    # Caminho completo do arquivo
    caminho_arquivo = os.path.join(pasta_origem, arquivo)

    # Verifica se é um arquivo (e não uma pasta)
    if os.path.isfile(caminho_arquivo):

        # Separa o nome da extensão
        nome, extensao = os.path.splitext(arquivo)

        # Remove o ponto da extensão e transforma em maiúsculo
        extensao = extensao[1:].upper()

        # Se o arquivo não tiver extensão
        if extensao == "":
            extensao = "OUTROS"

        # Cria a pasta correspondente à extensão (ex: PDF, JPG)
        pasta_extensao = os.path.join(pasta_destino, extensao)

        # Se a pasta ainda não existir, cria ela
        if not os.path.exists(pasta_extensao):
            os.makedirs(pasta_extensao)

        # Move o arquivo para a pasta correspondente
        shutil.move(caminho_arquivo, os.path.join(pasta_extensao, arquivo))

        # Mostra o que foi feito
        print(f"Movido: {arquivo} → {extensao}/")

print("✅ Organização concluída!")
``
```
