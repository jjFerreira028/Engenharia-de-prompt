```python
# ==========================================
# CONSULTA DE CEP USANDO API VIA CEP
# ==========================================

# Importa a biblioteca requests, usada para fazer requisições HTTP
import requests

# Solicita ao usuário que digite um CEP
cep = input("Digite o CEP (somente números): ")

# Monta a URL da API do ViaCEP com o CEP informado
url = f"https://viacep.com.br/ws/{cep}/json/"

try:
    # Faz a requisição GET para a API
    resposta = requests.get(url)

    # Converte a resposta para formato JSON (dicionário em Python)
    dados = resposta.json()

    # Verifica se o CEP é inválido (a API retorna 'erro' nesse caso)
    if "erro" in dados:
        print("❌ CEP não encontrado!")
    else:
        # Exibe as informações do endereço
        print("\n✅ Endereço encontrado:")
        print(f"Rua: {dados['logradouro']}")
        print(f"Bairro: {dados['bairro']}")
        print(f"Cidade: {dados['localidade']}")
        print(f"Estado: {dados['uf']}")

except requests.exceptions.RequestException:
    # Trata erros de conexão (ex: sem internet)
    print("⚠️ Erro ao conectar com a API. Verifique sua internet.")
```
